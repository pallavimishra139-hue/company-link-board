```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#pulls stock price data

tickers = {
    "Zara (Inditex)": "ITX.MC",
    "H&M": "HM-B.ST",
    "ASOS": "ASC.L",
    "LVMH": "MC.PA",
    "Hermès": "RMS.PA",
    "Kering": "KER.PA"
}

data = {}
for name, ticker in tickers.items():
    data[name] = yf.download(ticker, period="5y")
    print(f"{name}: {len(data[name])} days of data pulled")
```

```python
#pull company's financial statements

tickers = {
    "Zara (Inditex)": "ITX.MC",
    "H&M": "HM-B.ST",
    "ASOS": "ASC.L",
    "LVMH": "MC.PA",
    "Hermès": "RMS.PA",
    "Kering": "KER.PA"
}

financials = {}
for name, ticker in tickers.items():
  stock = yf.Ticker(ticker)
  financials[name] = {"income_statement" : stock.financials,
                      "balance_sheet" : stock.balance_sheet
                      }
  print(f"{name}: financial data pulled")
```

```python
financials["LVMH"]["income_statement"]
```

```python
#creates the gross profit formulaf or companies based in the varibales from the table and defines it first

gross_margins = {}


for name in tickers.keys():
    income = financials[name]["income_statement"]
    revenue = income.loc["Total Revenue"]
    gross_profit = income.loc["Gross Profit"]
    margin = gross_profit / revenue * 100
    gross_margins[name] = margin
    print(f"{name} gross margin by year:")
    print(margin)
    print()
```

```python
#similar format but defines how to solve for opearting margin instead

operating_margins = {}

for name in tickers.keys():
  income = financials[name]["income_statement"]
  revenue = income.loc["Total Revenue"]
  operating_income = income.loc["Operating Income"]
  margin = (operating_income)/revenue * 100
  operating_margins[name] = margin
  print(f"{name} operating margin by year")
  print(margin)
  print()
```

```python
print(financials["LVMH"]["income_statement"].index.tolist())
```

```python
#inventory turnover fire fire

inventory_turnover = {}

for name in tickers.keys():
  income = financials[name]["income_statement"]
  balance = financials[name]["balance_sheet"]

  cogs = income.loc["Cost Of Revenue"]
  inventory = balance.loc["Inventory"]

  avg_inventory = inventory.rolling(window=2).mean() #averages the inventory so its more relaible & not spiky

  turnover = cogs / avg_inventory
  inventory_turnover[name] = turnover
  print(f"inventory turnover by year:")
  print(turnover)
  print()


```

```python
sga_pct = {}

for name in tickers.keys():
  income = financials[name]["income_statement"]
  revenue = income.loc["Total Revenue"]

  if name == "Kering":
        print(f"{name}: SG&A not separately disclosed, skipping")
        continue

  sga = income.loc["Selling General And Administration"]
  sga_pct[name] = sga / revenue * 100
  print(f"{name} SG&A %: \n{sga_pct[name]}")
  print()
```

```python
ebitda_margin = {}
net_margin = {}
marketing_pct = {}

for name in tickers.keys():
    income = financials[name]["income_statement"]
    revenue = income.loc["Total Revenue"]

    try:
     ebitda = income.loc["EBITDA"]
     ebitda_margin[name] = ebitda / revenue * 100
     print(f"{name} EBITDA margin %: \n{ebitda_margin[name]}\n")
    except Exception as e:
        print(f"{name}: EBITDA failed — {type(e).__name__}: {e}\n")
```

```python
#isnt printing other companies debug later

net_margin = {}

for name in tickers.keys():
    income = financials[name]["income_statement"]
    revenue = income.loc["Total Revenue"]

    try:
        net_income = income.loc["Net Income"]
        net_margin[name] = net_income / revenue * 100
        print(f"{name} Net margin %: \n{net_margin[name]}\n")
    except Exception as e:
        print(f"{name}: Net margin failed — {type(e).__name__}: {e}\n")
```

```python
marketing_pct = {}

for name in tickers.keys():
    income = financials[name]["income_statement"]
    revenue = income.loc["Total Revenue"]

    try:
        marketing = income.loc["Selling And Marketing Expense"]
        marketing_pct[name] = marketing / revenue * 100
        print(f"{name} Marketing %: \n{marketing_pct[name]}\n")
    except KeyError:
        print(f"{name}: Marketing expense row not found, skipping\n")
```

```python
#start of charts

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

for name, margin in gross_margins.items():
    plt.plot(margin.index, margin.values, marker='o', label=name)

plt.title("Gross Margin: Fast Fashion vs. Luxury (2021-2026)")
plt.xlabel("Year")
plt.ylabel("Gross Margin (%)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.gca().invert_xaxis()
plt.tight_layout()
plt.show()
```

This project compares the financial performance of Fast Fashion vs. Luxury Fashion. I used H&M, ASOS, and Zara for fast fashion and used Kering, Hermes, and LVMH for luxury fashion. As for metric, I used gross margins, operating margins, inventory turnover, SG&A, EBITDA, marketting expense, and net profil margins.

---



Findings

---



*   **Gross Margin Analysis:**
There is a clear difference between the gross margins between fast fashion companies and luxury fashion companies. Luxury brands run gross margins of roughly 66-76%, compared to 40-58% for fast fashion. This indicates that luxury brands have a higher degree of pricing power where they can charge above market pricing and still not take a hit to their revenues. Hermes is a greta example because it holds steady at Hermes holds steady at 70-72% which shows pandemic and external macroeconomic factors don't affect sales as much as they affect fast fashion companies. This can be due to the brand loyalty that Hermes creates for it's customers.  



*   **Operating Margin Analysis**:
Operating margins help tell us what gross margin doesn't. Hermes leads again with turning 41-43% of their revenue into operating profit. LVMH is  relatively stable at around 21-28%. Kering is the most important finding as it dropped from ~28%(2023) to ~11% in 2026. Kering's gross margin still remains in the 70s. This gap reveals that Kering has been spending heavily on things like restructuring and marketing.
Meanwhile, Zara is the only one with similar operating margins like the luxury companies. H&M has gone up from 3% to ~8%. As for ASOS, it runs negative in most years, confirming real operational struggles beneath a superficially decent gross margin.



*   **Inventory Turnover**: This is where fast fashion companies dominate. Trying to keep up with current trends leads to quick inventory turnover in order to drive frequent sales. Turnover in fast fashion companies is 2.5x to 5x higher than luxury brands. Luxury brands, on the other hand, utilize scarcity marketing. Instead of bringing in new trends every season, they stick to differentiated, signature products that hold their value and appeal over time.


*   **SG&A Analysis**: People assume that luxury brands spend more money to protect and brand themselves against competitors. Howevver, the opposite is observed in the graph above. Zara runs by far the lowest SG&A of any company in the dataset, at just 9-10% of revenue. Additionally, Hermès keeps SG&A to 23-24%, notably lower than LVMH's 42-44%, despite Hermès having the strongest brand and highest margins of the group. This suggests Hermès doesn't need to spend heavily to sustain demand and the name of the brand sells itself. The real surprise is on the fast fashion side: H&M and ASOS both run SG&A in the 37-48% range, comparable to or even higher than LVMH. H&M's SG&A jumped sharply in the most recent year (37% to 47%), and ASOS has hovered near 43-48% throughout. Combined with their thin or negative operating margins seen earlier, this shows H&M and ASOS are spending at a near-luxury level on overhead and marketing without luxury-level pricing power to support it.
*(Note: Kering's SG&A isn't separately disclosed in the available data, so it's excluded from this comparison)*



*   **EBITDA Analysis**: EBITDA margin shows how profitable a company is before depreciation and amortization. Hermès leads at 46–49%, followed by LVMH at 32–35%. Kering declined sharply from about 34% to 21%, showing that its problems are operational, not just accounting-related. Zara stands out in fast fashion with a strong 25–29% margin, closer to LVMH than H&M or ASOS. H&M improved from 13% to 17.5%, while ASOS remained mostly negative.


*   **Net Profit Margin**: Net margin shows how much profit a company keeps after all expenses. Kering's collapse is the sharpest trend in the dataset: net margin fell from 17.8% (2022) to just 0.5% (2025), nearly breaking even. This mirrors its operating margin decline almost exactly, confirming the Gucci slump has hit Kering's actual bottom line, not just its operating results. ASOS was the weakest performer, with mostly negative net margins. Zara was the strongest fast-fashion company across the different profitability measures.


*   **Marketing Expense as % of Revenue**: The biggest takeaway is that higher marketing spending does not necessarily lead to higher profitability. LVMH spends heavily on marketing to maintain its luxury positioning, while Zara achieves strong performance while spending very little.

This also shows that different business models require different marketing strategies. Luxury companies use marketing to protect brand value and exclusivity, while fast-fashion companies generally focus more on speed, pricing, and product turnover. Overall, Zara is the most interesting outlier since it spends the least on marketing but remains one of the strongest performers.

Luxury and fast fashion brands are different models that utilise different business models to gain profit. Luxury protects pricing power and brand scarcity at the cost of speed and inventory efficiency; meanwhile, fast fashion sacrifices margin for volume and turnover speed. The most important finding in this analysis isn't the static gap between the two groups, but the trend within luxury itself. On the other hand, Meanwhile, Zara's consistent strength across every fast fashion metric shows that even in a lower-margin model, operational excellence can meaningfully separate winners from laggards.
